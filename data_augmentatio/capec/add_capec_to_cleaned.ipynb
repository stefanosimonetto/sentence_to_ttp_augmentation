```python
import json

# Assume your original data is loaded as `data`
# data = {"sentence": {...}, "labels": {...}}

def technique_url(tech_id: str) -> str:
    """
    Build MITRE ATT&CK technique URL.
    Handles sub-techniques correctly.
    """
    if "." in tech_id:
        base, sub = tech_id.split(".")
        return f"https://attack.mitre.org/techniques/{base}/{sub}"
    return f"https://attack.mitre.org/techniques/{tech_id}"

converted = []

for idx, sentence in data["sentence"].items():
    labels = data["labels"].get(idx, [])
    titles = data["titles"].get(idx, [])
    for tech_id in labels:
        record = {
            "source_type": "capec",
            "source_name": "CAPEC-derived",
            "relationship_type": "uses",
            "relationship_description": sentence.strip(),
            "technique_id": tech_id,
            "technique_url": technique_url(tech_id),
            "direction": "outgoing",
        }
        converted.append(record)

print(f"Generated {len(converted)} relationship objects")
```

```python
with open("../../data_augmentatio_stefano/mitre/cleaned_capec.json", "w") as f:
    json.dump(cleaned + converted, f, indent=4)
```
