```python
import pandas as pd

PATH = "/home/simonettos/thijs/classification/classification/tram_bosch_test_best_model_augmented.txt"   # or .csv

df = pd.read_csv(PATH, index_col=0)

metrics = ["precision_mean", "recall_mean", "f1_mean"]
for c in metrics:
    df[c] = pd.to_numeric(df[c], errors="coerce")

summary = (
    df.groupby(["model", "dataset"], as_index=False)
      .agg(
          n_runs=("f1_mean", "size"),

          precision_mean=("precision_mean", "mean"),
          precision_std=("precision_mean", "std"),

          recall_mean=("recall_mean", "mean"),
          recall_std=("recall_mean", "std"),

          f1_mean=("f1_mean", "mean"),
          f1_std=("f1_mean", "std"),
      )
      .sort_values(["dataset", "model"])
)

print(summary)
# Optional: save
summary.to_csv("/home/simonettos/thijs/classification/classification/model_dataset_means.csv", index=False)
```
